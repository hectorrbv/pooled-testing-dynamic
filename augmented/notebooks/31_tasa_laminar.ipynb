## 1. La tasa laminar exacta

$q=1/2$, $u=1$, población grande.



$c^*(G)$ es el mejor cociente $\mathbb E[\text{acreditados}]/\mathbb E[\text{pruebas}]$ de una estrategia que trabaja un solo bloque. Con $n\ge BG$, $V_{\rm lam}\le c^*(G)\,B$ y $V_{\rm lam}/B\to c^*(G)$. Vale $1/2,\ 2/3,\ 9/13,\ 14/19$ y llega a 0.869 en $G=96$. El potencial de Francisco da 2; el techo, $\log_2(G+1)$.

## 2. El óptimo laminar exacto se acerca a $c^*(G)$

18 casos exactos, $G\in\{2,3,4\}$, $B\le12$, $n=BG$.



Ningún caso supera $c^*(G)\,B$. Con $G=2$ y $B=12$ el óptimo vale 0.657 por prueba, contra $2/3$.

## 3. La regla óptima

$G=8$. Cada casilla dice cuántas personas del átomo se prueban.



Abrir el grupo más grande posible. Si un átomo tiene más de la mitad infectada, abandonarlo; si no, probar una parte de él, de alrededor de la mitad en átomos chicos. Es una regla de cociente, como la que pidió Francisco el 22 de septiembre, y es óptima con $q$ homogéneo y población grande.

## 4. General contra laminar, por prueba



Söderberg–Shapiro acredita 0.833 por prueba con $G=7$; el laminar, 0.768. El detector de 4 no le gana al laminar. El diseño del loop 1 ($n=16$, $B=4$, $G=8$) acredita 3.378 contra el óptimo laminar exacto 2.865. Si la tasa de Lindström valiera ya en $G\approx10$, la separación empezaría ahí.

## 5. Otros $q$



Con $q=0.9$ la tasa laminar vale 4.21 por prueba en $G=16$ y 4.81 en $G=48$; con $q=1/2$, 0.81 en $G=16$. En todos los $q$ la curva se aplana; lo que cambia es el nivel. La cota del precio por información alcanza su máximo en $q\approx0.75$, donde el techo cambia de régimen: no decide si $q=1/2$ es el peor caso.

## 6. El techo sin el $+1$

Hoja por hoja, $m(h)\le\log_2(1/P(h))$. Al promediar, $\mathbb E[m]\le H(\text{historial})\le\log_2K\le B\log_2(G+1)$. Con $q$ general: $\mathbb E[m]\le B\log_2(G+1)/\log_2(1/q)$ y $\le BG$.

En las 24 celdas de la malla, los despejados quedan bajo $\log_2K$ con margen mínimo 0.50. Derivación por validar con A.
