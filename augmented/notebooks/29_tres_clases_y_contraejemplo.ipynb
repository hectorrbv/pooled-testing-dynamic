## 1. Binario gana al laminar
Testigo: $n=3$, $B=2$, $G=3$, $u\equiv1$, $q=(0.9,0.7,0.7)$.









### La mayor separación: laminar/binario tiende a $9/10$

$n=4$, $B=G=2$, $q=(s,1/2,1/2,s)$, $u=(1,3,3,2)$, con $19/20\le s<1$. 



$V_{\rm bin}=\tfrac52\,s(1+s)$ y $V_{\rm lam}=3+\tfrac32\,s^2$. En $s=0.99$: binario 4.92525, laminar 4.47015. 

## 2. Laminar y general





## Tres pruebas identifican a cuatro personas

$n=4$, $B=3$, $G=3$, $q=1/2$, $u=1$.









La matriz identifica a todos. Como cada persona tiene probabilidad $1/2$ de estar sana, identifica $4\times1/2=2$ sanos en promedio.

Con las mismas tres pruebas, el óptimo laminar identifica 1.75 sanos en promedio. Probar tres personas individualmente da 1.5. Exigir laminaridad pierde el 12.5% de la utilidad general, incluso eligiendo la mejor política laminar.

La pregunta siguiente es si esa pérdida aumenta al construir diseños mayores.

## 3. Lo que muestra la malla

$n=7$, $B=4$, $G\in\{2,3,4,5\}$; cuatro de los 32 casos exactos.



Con $G=3$ la pérdida laminar es 21.2% del óptimo general. 

## 4. Lo que ya acota la ventaja



Ninguna política acredita más de $B\log_2(G+1)$ sanos en promedio. 

## Una matriz de 15 personas y 9 pruebas

$n=15$, $B=9$, $G=7$, $q=1/2$, $u=1$.

El diseño de [Söderberg y Shapiro (1963, §3)](https://www.jstor.org/stable/2312835) distingue los $32\,768$ perfiles. Su grupo más grande tiene siete personas. Identifica a todos y alcanza el óptimo general: $15/2=7.5$ sanos en promedio.



El óptimo laminar, calculado con Bellman exacto, es $203479/32768\approx6.2097$. Verificamos su política sobre los $32\,768$ perfiles, respetando laminaridad y $B=9$ en cada trayectoria.

Exigir laminaridad reduce la utilidad un 17.2% en este caso. Por unidad de presupuesto, $V/B\approx0.690$ para laminar y $0.833$ para la matriz. Esta separación finita todavía no demuestra crecimiento $\log G$.

## 5. Mejoró el diseño; queda limitada su escala

$n=16$, $B=4$. Comparamos el diseño inicial con el encontrado por la búsqueda, no con el óptimo general.



El diseño encontrado gana en los cuatro casos. Pero trabaja en bloques separados de hasta seis personas y prueba como máximo cinco juntas. Permitir $G=8$ o $G=16$ no cambia sus opciones. La meseta es una restricción de la búsqueda; no refuta el crecimiento $\log G$.

Los seis objetivos del loop, incluidos los de $B=8$, coinciden exactamente con una evaluación independiente de los $2^{16}$ perfiles.

Siguiente experimento: permitir grupos que combinen bloques y revisar la primera decisión que mejore el valor. Después, hacer crecer $n$, $B$ y el tamaño efectivo de los grupos.

## 6. Heurísticas: ¿compensa planificar más?

$n=50$, $B=5$, $G\in\{3,5\}$; $q_i$ uniforme en $[0,1]$ y $u_i$ uniforme en $\{1,2,3\}$. Las tres reglas usan la misma partición fija y la misma selección de pruebas permitidas dentro de ella. La utilidad se estima por Monte Carlo sobre los mismos perfiles.

Greedy elige el cobro inmediato. Rollout compara una primera prueba suponiendo que después seguirá greedy. La regla nueva explora varias decisiones hasta cuatro pruebas. Ambas vuelven a decidir tras cada resultado.



Rollout mejora a greedy un 2.01% en $G=3$ y un 5.20% en $G=5$. La regla nueva añade 0.16% y 0.34%, pero los intervalos aproximados del 95% incluyen cero. Cuesta 3.8 y 9.7 veces más en estas mediciones.

Mantendría rollout como referencia práctica. No hay evidencia clara de que la regla nueva lo mejore aquí; tampoco hemos demostrado que rindan igual. Esta comparación no mide el precio de exigir laminaridad.
